# Remove Duplicates

Step 1: Identify duplicate rows

```python
duplicates = restaurants_df.duplicated(keep=False)

restaurants_df[duplicates].sort_values(by='name')
```

Step 2: Review duplicate records

```python
display(restaurants_df[duplicates].groupby('name').count())
```

Step 3: Remove duplicate rows (keeping the first occurrence)

```python
restaurants_cleaned_df = restaurants_df.drop_duplicates(keep='first')
```

Step 4: Verify duplicate removal

```python
display(restaurants_df.shape)
display(restaurants_cleaned_df.shape)
```

```python
duplicates = restaurants_cleaned_df.duplicated(keep=False)

restaurants_cleaned_df[duplicates].sort_values(by='name')
```

```python
display(restaurants_cleaned_df[duplicates].groupby('name').count())
```

# Cleaning the Rate Column

Step 1: Checking How many NEW values there in rate column

```python
a = restaurants_cleaned_df[restaurants_cleaned_df['rate'] =='NEW']
a
```

Step 2: Replacing '/5' and 'NEW'

```python
restaurants_cleaned_df['rate'] = restaurants_cleaned_df['rate'].str.replace("/5","").str.strip().replace("NEW",np.nan)
```

Step 3: Make Sure Replaced values updated

```python
restaurants_cleaned_df['rate'].unique()
```

```python
restaurants_cleaned_df['rate'].isnull().sum()
```

Step 4: Checking Data Type of Rate column

```python
restaurants_cleaned_df['rate'].dtype
```

Step 5: Chnaging Data Type

```python
restaurants_cleaned_df['rate'] = pd.to_numeric(restaurants_cleaned_df['rate'], errors='coerce')
```

Step 6: Make Sure Data Type changed

```python
restaurants_cleaned_df['rate'].dtype
```

# Price Segments New Column

```python
restaurants_cleaned_df['approx_cost(for two people)'].max()
```

```python
restaurants_cleaned_df['approx_cost(for two people)'].min()
```

```python
restaurants_cleaned_df['approx_cost(for two people)'].unique()
```

Column Creation

```python
conditions = [
    restaurants_cleaned_df['approx_cost(for two people)'] <= 500,
    (restaurants_cleaned_df['approx_cost(for two people)'] > 500) & (restaurants_cleaned_df['approx_cost(for two people)'] <= 1000),
    restaurants_cleaned_df['approx_cost(for two people)'] > 1000
]

choices = ['Budget', 'Mid', 'Premium']

restaurants_cleaned_df['price_category'] = np.select(conditions, choices, default='Unknown')
```

Verify

```python
restaurants_cleaned_df['price_category'].value_counts()
```

```python
restaurants_cleaned_df[['approx_cost(for two people)','price_category']].head()
```

# Rating Segment New Column

Column Create

```python
conditions = [
    restaurants_cleaned_df['rate'] < 3.0,
    (restaurants_cleaned_df['rate'] >= 3.0) & (restaurants_cleaned_df['rate'] < 4.0),
    restaurants_cleaned_df['rate'] >= 4.0  
]

choices = ['Poor', 'Good', 'Excellent']

restaurants_cleaned_df['rating_category'] = np.select(
    conditions,
    choices,
    default='Unknown'
)
```

Verify

```python
restaurants_cleaned_df['rating_category'].value_counts()
```

```python
restaurants_cleaned_df[['rate','rating_category']].head()
```

Save the Cleaned Dataset
